```python
# Try to create a Series and DataFrame with str, tuple, list[list],list[tuple],Set,dict
import pandas as pd
```

```python
# List to Data frame
lst1=list(range(1,11))
print('Orginal List:',lst1)
lst_df=pd.DataFrame(lst1)

print("List to Data frame is:",lst_df)
```

```python
# String to DF
str1='Hello Python'
str_list=list(str1)
print('String to List:',str_list)
str_df=pd.DataFrame(str_list, columns=["Char"])

print('String to Data Frame:',str_df)
```

```python
# take user names as user inouts

user_input=input('Enter student names seprated by comma:')

name_df=pd.DataFrame(user_input.split(","), columns=["Student Name"])

print("Student names in Data frame:",name_df)
```

```python
# touple to data frame conversion

tp1=tuple(range(21,41))
tuple_df=pd.DataFrame(tp1, columns=["Tuple Elements"])

print('Tuple to DF:\n',tuple_df)
```

```python
# list[list] to data frame

listoflist=[['GS','PG','RS'],[33,33,24],['Sangola', 'Solapur','Pune']]
list_df2=pd.DataFrame(listoflist)

print(list_df2)
```

```python
# list[tuple] to data frame
lst_tp=[1,2,3,('A','B','C'),[4,5,'Six'],'End']

lst_df3=pd.DataFrame(lst_tp,columns=['List Elements'])
lst_df3
```

```python
# conver the disctonry to Data frame
dict_1={'Name':'Gorakh','Age':33,'City':'Pune','Skills':'Python'}
dict_df=pd.DataFrame(dict_1,index=[1])
print(dict_df)
#
dist1={'Name':['Gorakh','Ajay','Sunil'],'Age':[33,34,20]}
dist_df=pd.DataFrame(dist1)
print(dist_df)
```

```python
# set to data frame
set1={'A','B',"c",'Test'}
set_df=pd.DataFrame(set1)
print(set_df)
#
```
